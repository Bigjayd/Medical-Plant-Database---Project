# Data Cleaning & Standardization

```python
def clean_data(df, source_name):
    """
    Cleans the input DataFrame by removing duplicates and handling missing values.
    
    Parameters:
    df (pd.DataFrame): The input DataFrame to be cleaned.
    source_name (str): The name of the data source.
    
    Returns:
    pd.DataFrame: The cleaned DataFrame.
    """
    # Remove duplicates
    df = df.drop_duplicates()

    # Fix column dtypes if necessary (e.g., convert to string)
    df['Family'] = df['Family'].astype("string")

    return df

def standardize_data(df, source_name):
    """
    Standardizes the input DataFrame by correcting column names.
    
    Parameters:
    df (pd.DataFrame): The input DataFrame to be standardized.
    
    Returns:
    pd.DataFrame: The standardized DataFrame.
    """

    # Rename Column Names to Remove Spaces and Special Characters
    df.columns = df.columns.str.replace(" ", "_").str.replace("(", "").str.replace(")", "").str.replace("_FNFNUM", "").str.lower()

    # Clean plant_species column by removing extra spaces and converting to lowercase
    df['plant_species'] = df['plant_species'].str.strip().str.replace(" (*)", "").str.lower()
    return df

# Fill missing Family information based on Plant Species
FAMILY_MAP = {
    "Senna podocarpa": "Fabaceae",
    "Senna alata": "Fabaceae",
    "Garcinia kola": "Clusiaceae",
    "Ficus exasperata": "Moraceae",
    "Carica papaya": "Caricaceae",
    "Xylopia aethiopica": "Annonaceae",
    "Rauvolfia vomitoria": "Apocynaceae",
    "Nicotiana tabacum": "Solanaceae",
    "Curculigo pilosa": "Hypoxidaceae",
    "Vernonia amygdalina": "Asteraceae",
    "Plumbago zeylanica": "Plumbaginaceae",
    "Zingiber officinale": "Zingiberaceae",
    "Newbouldia laevis": "Bignoniaceae",
    "Spondias mombin": "Anacardiaceae",
    "Phyllanthus amarus": "Phyllanthaceae",
    "Picralima nitida": "Apocynaceae",
    "Culcasia scandens": "Araceae",
    "Momordica charantia": "Cucurbitaceae",
    "Capsicum frutescens": "Solanaceae",
    "Ocimum gratissimum": "Lamiaceae",
    "Citrus aurantiifolia": "Rutaceae",
    "Citrus medica": "Rutaceae",
    "Jatropha curcas": "Euphorbiaceae",
    "Citrullus colocynthis": "Cucurbitaceae",
    "Ricinus communis": "Euphorbiaceae",
    "Allium sativum": "Amaryllidaceae",
    "Allium ascalonicum": "Amaryllidaceae",
    "Crassocephalum crepidioides": "Asteraceae",
    "Aframomum melegueta": "Zingiberaceae",
}
def fill_missing_family(df, source_name):
    """
    Fills missing values in the 'family' column based on the 'plant_species' column using a predefined mapping.
    
    Parameters:
    df (pd.DataFrame): The input DataFrame with potential missing values in the 'family' column.
    
    Returns:
    pd.DataFrame: The DataFrame with missing 'family' values filled where possible.
    """
    for species, family in FAMILY_MAP.items():
        df.loc[df['plant_species'] == species.lower(), 'family'] = family

    # Handle any remaining missing values in the 'family' column
    df['family'] = df['family'].fillna("Unknown")
    return df 

# Standardize and complete plant part information
PLANT_PART_MAP = {
    "leaf": "Leaf",
    "leaves": "Leaf",
    "Leaf": "Leaf",
    "stem": "Stem",
    "Stem": "Stem",
    "root": "Root",
    "Root": "Root",
    "rhizome": "Rhizome",
    "Rhizome": "Rhizome",
    "fruit": "Fruit",
    "Fruit": "Fruit",
    "seed": "Seed",
    "Seed": "Seed",
    "bark": "Bark",
    "Bark": "Bark",
    "flower": "Flower",
    "Flower": "Flower",
    "whole plant": "Whole plant",
    "Whole plant": "Whole plant",
    "bulb": "Bulb",
    "Bulb": "Bulb",
    "Essential oil": "Essential oil",
    "Seed Essent. Oil": "Essential oil",
    "Rhizome Essent. Oil": "Essential oil",
    "Root Essent. Oil": "Essential oil",
    "Fruit Essent. Oil": "Essential oil",
    "Seed Oil": "Essential oil",
    "Oil": "Essential oil",
    "pericarp": "Pericarp",
    "plant": "Whole plant",
}
def standardize_plant_part(df, source_name):
    """
    Standardizes the 'plant_part' column in the DataFrame based on a predefined mapping.
    
    Parameters:
    df (pd.DataFrame): The input DataFrame with potential variations in the 'plant_part' column.
    
    Returns:
    pd.DataFrame: The DataFrame with standardized 'plant_part' values.
    """
    df['plant_part'] = df['plant_part'].map(PLANT_PART_MAP).fillna(df['plant_part'])

    # Handle any remaining missing values in the 'plant_part' column
    df['plant_part'] = df['plant_part'].fillna("Unknown")
    return df

# Clean and Enrich Phytochemical Names
def clean_phytochemical_names(df, source_name):
    """
    Cleans and standardizes the 'phytochemical_name' column in the DataFrame.
    
    Parameters:
    df (pd.DataFrame): The input DataFrame with potential variations in the 'phytochemical_name' column.
    
    Returns:
    pd.DataFrame: The DataFrame with cleaned and standardized 'phytochemical_name' values.
    """
    # Remove extra spaces, special characters, and convert to lowercase
    df['phytochemical'] = df['phytochemical'].str.strip().str.replace(r"[^a-zA-Z0-9\s]", "", regex=True).str.lower()
    
    # Handle any remaining missing values in the 'phytochemical_name' column
    df['phytochemical'] = df['phytochemical'].fillna("Unknown")
    
    return df

# Clean and Standardize the combined Data
df_combined = clean_data(df_combined, "Combined")
df_combined = standardize_data(df_combined, "Combined")
df_combined = fill_missing_family(df_combined, "Combined")
df_combined = standardize_plant_part(df_combined, "Combined")
df_combined = clean_phytochemical_names(df_combined, "Combined")

# Clean and Standardize the NMPPDB Data
df_nmppdb = clean_data(df_nmppdb, "NMPPDB")
df_nmppdb = standardize_data(df_nmppdb, "NMPPDB")
df_nmppdb = fill_missing_family(df_nmppdb, "NMPPDB")
df_nmppdb = standardize_plant_part(df_nmppdb, "NMPPDB")
df_nmppdb = clean_phytochemical_names(df_nmppdb, "NMPPDB")

# Clean and Standardize the IMPPAT Data
df_imppat = clean_data(df_imppat, "IMPPAT")
df_imppat = standardize_data(df_imppat, "IMPPAT")
df_imppat = fill_missing_family(df_imppat, "IMPPAT")
df_imppat = standardize_plant_part(df_imppat, "IMPPAT")
df_imppat = clean_phytochemical_names(df_imppat, "IMPPAT")

# Clean and Standardize the DrDuke Data
df_drduke = clean_data(df_drduke, "DrDuke")
df_drduke = standardize_data(df_drduke, "DrDuke")
df_drduke = fill_missing_family(df_drduke, "DrDuke")
df_drduke = standardize_plant_part(df_drduke, "DrDuke")
df_drduke = clean_phytochemical_names(df_drduke, "DrDuke")

print("Data cleaning and standardization complete.")
display(df_combined.head())
```

# Build Master Dataset

```python
print("\n" + "=" * 70)
print("STEP 7: Building prioritized master dataset...")
print("=" * 70)

# Priority order: Dr Duke (richest) > NMPPDB > IMPPAT > Combined
# We concatenate and then deduplicate intelligently

df_master = pd.concat([df_drduke, df_nmppdb, df_imppat, df_combined], ignore_index=True)
print(f"Raw concatenated rows: {len(df_master)}")

# Remove completely empty phytochemical rows
df_master = df_master.dropna(subset=["phytochemical"])
df_master = df_master[df_master["phytochemical"].str.len() > 1]
print(f"Rows after removing empty phytochemical entries: {len(df_master)}")

# Create uniqueness key based on plant_species, plant_part, and phytochemical
df_master["unique_key"] = (
    df_master["plant_species"].astype(str) + "||" +
    df_master["plant_part"].astype(str) + "||" +
    df_master["phytochemical"].astype(str).str.lower()
)

# Prefer rows that have more filled information.
df_master['info_score'] = (
    df_master['family'].notna().astype(int) * 2 +
    df_master['phytochemical_class'].notna().astype(int) * 3 +
    df_master['pharmacological_activities_compound'].notna().astype(int) * 3 +
    df_master['dosage'].notna().astype(int) +
    df_master['toxicity'].notna().astype(int) +
    df_master['references'].notna().astype(int) +
    df_master['ethnobotanical_uses'].notna().astype(int) + 
    df_master['biological_activity_plant_extract'].notna().astype(int)
)

# Sort so best rows come first, then drop duplicates based on unique_key, keeping the first (best) entry
df_master = df_master.sort_values(by=["unique_key", "info_score"], ascending=[True, False])
df_master = df_master.drop_duplicates(subset=["unique_key"], keep="first")
df_master = df_master.drop(columns=["unique_key", "info_score"])

# Final family fill (in case any remain)
df_master = fill_missing_family(df_master, "Master")

print("Master dataset creation complete.")
print(f"Final master dataset rows: {len(df_master)}, Columns: {df_master.shape[1]}, Rows: {df_master.shape[0]}")
print("First 5 rows of the master dataset:")
display(df_master.head())
```

# Quality Report

```python
# ============================================================
# 8. QUALITY REPORT
# ============================================================
print("\n" + "=" * 70)
print("STEP 8: Quality Report")
print("=" * 70)

print("\n--- Missing values in Master ---")
missing = df_master.isnull().sum()
missing_pct = (missing / len(df_master) * 100).round(1)
quality = pd.DataFrame({
    "Missing_Count": missing,
    "Missing_%": missing_pct
})
print(quality)

print("\n--- Unique counts ---")
print(f"Unique Plant Species : {df_master['plant_species'].nunique()}")
print(f"Unique Phytochemicals: {df_master['phytochemical'].nunique()}")
print(f"Unique Families      : {df_master['family'].nunique()}")

print("\n--- Top 15 plants by number of compounds ---")
print(df_master["plant_species"].value_counts().head(15))

print("\n--- Source distribution in final master ---")
print(df_master["source_database"].value_counts())

print("\n--- Top 15 most frequent phytochemicals ---")
print(df_master["phytochemical"].value_counts().head(15))
```

# Basic Ananlysis Tables

```python
# ============================================================
# 9. BASIC ANALYSIS TABLES
# ============================================================
print("\n" + "=" * 70)
print("STEP 9: Generating summary tables...")
print("=" * 70)

# Compounds per plant
compounds_per_plant = (
    df_master.groupby("plant_species")["phytochemical"]
    .nunique()
    .reset_index(name="Unique_Compounds")
    .sort_values("Unique_Compounds", ascending=False)
)

# Family summary
family_summary = (
    df_master.groupby("family")
    .agg(
        Plants=("plant_species", "nunique"),
        Compounds=("phytochemical", "nunique"),
        Records=("phytochemical", "count")
    )
    .reset_index()
    .sort_values("Compounds", ascending=False)
)
print("\n--- Family Summary ---")
display(family_summary.head(10).reset_index(drop=True))

print("\n--- Compounds per Plant Summary ---")
display(compounds_per_plant.head(10).reset_index(drop=True))
```

# Export Results


```python
# ============================================================
# 10. EXPORT RESULTS
# ============================================================
print("\n" + "=" * 70)
print("STEP 10: Exporting results...")
print("=" * 70)

# Main clean master
master_path = os.path.join(OUTPUT_DIR, "Master_Phytochemical_Database_Clean.xlsx")
df_master.to_excel(master_path, index=False)
print(f"✓ Master clean table saved → {master_path}")

# Summary workbook
summary_path = os.path.join(OUTPUT_DIR, "Phytochemical_Summary_Statistics.xlsx")
with pd.ExcelWriter(summary_path, engine="openpyxl") as writer:
    compounds_per_plant.to_excel(writer, sheet_name="Compounds_per_Plant", index=False)
    family_summary.to_excel(writer, sheet_name="Family_Summary", index=False)
    quality.to_excel(writer, sheet_name="Missing_Values")
    df_master["plant_species"].value_counts().reset_index().to_excel(
        writer, sheet_name="Plant_Frequency", index=False
    )
    df_master["phytochemical"].value_counts().head(50).reset_index().to_excel(
        writer, sheet_name="Top_Phytochemicals", index=False
    )
print(f"✓ Summary statistics saved → {summary_path}")

# Also save CSV versions for easier use in other tools
df_master.to_csv(os.path.join(OUTPUT_DIR, "Master_Phytochemical_Database_Clean.csv"), index=False)
print(f"✓ CSV version saved")

# Export master dataset to XML for schema validation
xml_path = os.path.join(OUTPUT_DIR, "sample.xml")
try:
    df_master.to_xml(xml_path, index=False, root_name="dataset", row_name="record")
    print(f"✓ XML sample saved → {xml_path}")
except AttributeError:
    print("Pandas does not support to_xml in this environment. Install pandas >= 1.5 to enable XML export.")

print("\n" + "=" * 70)
print("PIPELINE COMPLETED SUCCESSFULLY")
print("=" * 70)
print(f"\nFinal master table: {len(df_master)} rows")
print(f"Unique plants     : {df_master['plant_species'].nunique()}")
print(f"Unique compounds  : {df_master['phytochemical'].nunique()}")
print(f"\nFiles written to: {OUTPUT_DIR}")
```
